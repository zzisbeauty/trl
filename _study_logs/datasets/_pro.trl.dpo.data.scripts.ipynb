```python
import json
from datasets import load_dataset  


dataPath = "/home/nvidia/Desktop/localdatasets/rl2-dpo-datas-from-hf-ultrafeedback_binarized"

# 加载数据集  
print("正在加载数据集...")  
dataset = load_dataset(  
    dataPath,  
    split="train"  
)

print(f"\n数据集总样本数: {len(dataset)}")  
print(f"\n数据集列名: {dataset.column_names}")  
print(f"\n数据集特征: {dataset.features}")  

# 查看前5个样本  
print("\n" + "="*80)  
print("前5个样本:")  
print("="*80)  
  
for i in range(min(5, len(dataset))):  
    print(f"\n样本 {i+1}:")  
    print("-" * 40)  
    sample = dataset[i]  
    for key, value in sample.items():  
        if isinstance(value, str) and len(value) > 100:  
            print(f"{key}: {value[:100]}... (截断)")  
        else:  
            print(f"{key}: {value}")  
  
# 保存完整的前10个样本到JSON文件  
print("\n" + "="*80)
print("保存前10个样本到 dataset_samples.json")  
samples_to_save = []
for i in range(min(10, len(dataset))):  
    samples_to_save.append(dataset[i]) 

# 已经存储了 demo 数据了，无需再次存储
# with open("_pro.dpo.data.scripts.dataset_samples.json", "w", encoding="utf-8") as f:  
#     json.dump(samples_to_save, f, ensure_ascii=False, indent=2)  
  
print("已保存到 dataset_samples.json")  
  
# 统计信息  
print("\n" + "="*80)  
print("数据集统计信息:")  
print("="*80)  
  
# 如果有 prompt 字段,统计长度  
if "prompt" in dataset.column_names:  
    prompt_lengths = [len(sample["prompt"]) for sample in dataset]  
    print(f"Prompt 平均长度: {sum(prompt_lengths) / len(prompt_lengths):.2f} 字符")  
    print(f"Prompt 最短长度: {min(prompt_lengths)} 字符")  
    print(f"Prompt 最长长度: {max(prompt_lengths)} 字符")  

print("\n完成!")
```
